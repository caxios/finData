```python
import findata
# 1. Import both functions explicitly
from findata.sec.api.api_10kq import list_documents, get_document_detail
```

```python
# 2. Get the list of all filings for AAPL
docs = list_documents('AAPL')
# Let's see what we got (this will show a list of dictionaries)
# Each dictionary has an 'accession_number', 'form_type', 'filing_date', etc.
print("Found filings:", len(docs["filings"]))
docs
```

```python
# 3. Grab the accession number of the very first one (the most recent filing)
my_accession_number = docs["filings"][0]["accession_number"]
print("Most recent accession number is:", my_accession_number)
```

```python
# 4. NOW you can get the specific document details!
document_detail = get_document_detail('AAPL', my_accession_number)
# Do something with your data!
# (For example, it contains things like 'business', 'risk_factors', 'mda')
print(document_detail.keys())
```
