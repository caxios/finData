```python
# Latest 10-K as a PDF — no URL typing, no 404s
findata.download_filing_pdf_for("AAPL", form_type="10-K", output_path="output/aapl_10k.pdf")

# A quarter's 10-Qs as a ZIP
findata.download_filing_pdf_for(
    "AAPL", form_type="10-Q",
    date_from="2024-01-01", date_to="2024-12-31",
    output_path="output/aapl_10q_2024.zip",
)

# Or: resolve real URLs first, then download
filings = findata.get_filings("AAPL", form_type="10-Q", count=3, parse_sections=False)
urls = [f["document_url"] for f in filings if f.get("document_url")]
findata.download_filing_pdf(urls, output_path="output/aapl_10q_bundle.zip")
```
